```python
#our function is called add. With this funtion you can add 2 numbers together. 
#The function takes 2 numbers as input and returns the sum of those numbers.
import numpy as np

def add(a, b):
    return int(np.add(a, b))

def list_of_sums(numbers):
    sums = []
    total = 0
    for num in numbers:
        total += num
        sums.append(total)
    return sums
```

```python
#Here are some examples of how to use the add function:
print(f'2 + 3 is {add(2, 3)}')
print(f'5 + 3 is {add(5, 3)}')
print(f'134 + 2 is {add(134, 2)}')
```

```python
print(list_of_sums([1, 2, 3, 4]))
```
