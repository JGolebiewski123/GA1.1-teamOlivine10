```python
import numpy as np
```

```python
def subtract_function(a,b):
    """
    Subtract 2 numbers

    Parameters:
    
    a: float
        First value

    b: float

        Second value
    """

    c = np.subtract(a,b)

    return c

```

```python
a = 10
b = 4

c = subtract_function(a,b)

print('The result of subtracting {} and {} is {}'.format(a,b,c))
```

```python
def distance_from_zero(x):
    if x >= 0:
        return x
    return subtract_function(0, x)


distance_from_zero(5)
```

```python
distance_from_zero(0)
```

```python
distance_from_zero(9)
```
