```python
import numpy as np
```

```python
def subtract_function(a,b):
    """
    Subtract 2 numbers

    Parameters:
    
    a: float
        First value

    b: float

        Second value
    """

    c = np.subtract(a,b)

    return c

```

```python
a = 10
b = 4

c = subtract_function(a,b)

print('The result of subtracting {} and {} is {}'.format(a,b,c))
```
